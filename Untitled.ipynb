```python
from pathlib import Path
import polars as pl

BASE_DIR = Path("data/CGMacros_dateshifted365/CGMacros")
BIO_PATH = BASE_DIR / "bio.csv"

NUM_COLS = [
    "Libre GL", "Dexcom GL", "HR", "Calories (Activity)", "METs",
    "Calories", "Carbs", "Protein", "Fat", "Fiber", "Amount Consumed"
]

arquivos_csv = sorted(BASE_DIR.glob("CGMacros-*/CGMacros-*.csv"))

lista_dfs = []
for csv_path in arquivos_csv:
    subject_id = int(csv_path.stem.split("-")[-1])
    df_temp = pl.read_csv(csv_path, infer_schema_length=10000)
    df_temp = df_temp.rename({c: c.strip() for c in df_temp.columns})
    if "Unnamed: 0" in df_temp.columns:
        df_temp = df_temp.drop("Unnamed: 0")
    
    cols_presentes = [c for c in NUM_COLS if c in df_temp.columns]
    
    df_temp = df_temp.with_columns([
        pl.lit(subject_id).alias("subject"),
        pl.col(cols_presentes).cast(pl.Float64, strict=False),
    ])
    lista_dfs.append(df_temp)

df = pl.concat(lista_dfs, how="diagonal_relaxed")
print(f"df carregado na memória: {df.shape}")
```

```python
# 1. Visão geral do tamanho e cobertura de participantes
print(f"Total de linhas (minutos): {df.height:,}")
print(f"Total de participantes   : {df['subject'].n_unique()}")

# 2. Porcentagem de valores válidos (não nulos) por coluna
preenchimento = df.select([
    (pl.col(c).is_not_null().mean() * 100).round(2).alias(c)
    for c in df.columns
])
print("\n--- Preenchimento (%) por coluna ---")
display(preenchimento)

# 3. Auditoria por participante (identifica quem tem falhas em sensores ou poucas refeições)
auditoria_sujeitos = (
    df.group_by("subject")
    .agg([
        pl.len().alias("total_minutos"),
        (pl.len() / 1440).round(1).alias("dias_gravados"),
        (pl.col("Libre GL").is_not_null().mean() * 100).round(1).alias("libre_pct"),
        (pl.col("Dexcom GL").is_not_null().mean() * 100).round(1).alias("dexcom_pct"),
        (pl.col("HR").is_not_null().mean() * 100).round(1).alias("hr_pct"),
        pl.col("Meal Type").is_not_null().sum().alias("qtd_refeicoes"),
    ])
    .sort("subject")
)
print("\n--- Resumo por participante (primeiros 10) ---")
display(auditoria_sujeitos.head(10))
```

```python
# 1. Estatísticas descritivas das variáveis contínuas
display(
    df.select(["Libre GL", "Dexcom GL", "HR", "Calories (Activity)", "METs"])
    .describe()
)

# 2. Distribuição das refeições por tipo (Breakfast, Lunch, Dinner, Snacks)
resumo_refeicoes = (
    df.filter(pl.col("Meal Type").is_not_null())
    .group_by("Meal Type")
    .agg([
        pl.len().alias("contagem"),
        pl.col("Calories").mean().round(1).alias("kcal_media"),
        pl.col("Carbs").mean().round(1).alias("carbs_media_g"),
        pl.col("Protein").mean().round(1).alias("prot_media_g"),
        pl.col("Fat").mean().round(1).alias("gord_media_g"),
        pl.col("Fiber").mean().round(1).alias("fibra_media_g"),
    ])
    .sort("contagem", descending=True)
)
print("--- Perfil nutricional médio por tipo de refeição ---")
display(resumo_refeicoes)

# 3. Correlação e diferença média entre Libre GL e Dexcom GL quando ambos estão ativos
ambos_ativos = df.filter(pl.col("Libre GL").is_not_null() & pl.col("Dexcom GL").is_not_null())
corr_cgm = ambos_ativos.select(pl.corr("Libre GL", "Dexcom GL")).item()
diff_media = ambos_ativos.select((pl.col("Dexcom GL") - pl.col("Libre GL")).mean()).item()

print(f"Correlação de Pearson (Libre vs Dexcom) : {corr_cgm:.4f}")
print(f"Viés médio (Dexcom - Libre)             : {diff_media:+.2f} mg/dL")
```

```python
df_prep = (
    df.with_columns([
        pl.col("Timestamp").str.to_datetime("%Y-%m-%d %H:%M:%S"),
        pl.mean_horizontal(["Libre GL", "Dexcom GL"]).alias("GL_mean"),
    ])
    .sort(["subject", "Timestamp"])  # ordenar ANTES de qualquer rolling/shift/interpolate
    .with_columns([
        # Só mantém o valor interpolado se houver leitura real a até 15 linhas (~15 min) de cada lado,
        # ou seja, buracos de até ~30 min
        pl.when(
            pl.col("GL_mean").forward_fill(limit=15).over("subject").is_not_null()
            & pl.col("GL_mean").backward_fill(limit=15).over("subject").is_not_null()
        )
        .then(pl.col("GL_mean").interpolate(method="linear").over("subject"))
        .otherwise(None)
        .alias("GL_interp"),

        (((pl.col("Timestamp") - pl.col("Timestamp").min().over("subject")).dt.total_minutes() // 1440) + 1)
        .cast(pl.Int32)
        .alias("study_day"),
    ])
)
```

```python
# Calcula estatísticas nos 90 minutos ANTERIORES (passado -> t0)
# e nos 90 minutos POSTERIORES (t0 -> +90 min usando shift(-90))
df_janelas = df_prep.with_columns([
    # --- Contexto no instante t0 e -90 minutos (PRÉ-refeição) ---
    pl.col("GL_interp").alias("GL_t0"),
    pl.col("GL_interp").rolling_mean(window_size=90, min_samples=20).over("subject").alias("GL_pre_mean_90m"),
    pl.col("GL_interp").rolling_std(window_size=90, min_samples=20).over("subject").alias("GL_pre_std_90m"),
    (pl.col("GL_interp") - pl.col("GL_interp").shift(90).over("subject")).alias("GL_delta_pre_90m"),
    pl.col("HR").rolling_mean(window_size=90, min_samples=20).over("subject").alias("HR_pre_mean_90m"),
    pl.col("METs").rolling_sum(window_size=90, min_samples=20).over("subject").alias("METs_pre_sum_90m"),

    # --- Resposta nos +90 minutos seguintes (PÓS-refeição) ---
    pl.col("GL_interp").shift(-90).rolling_mean(window_size=90, min_samples=30).over("subject").alias("GL_post_mean_90m"),
    pl.col("GL_interp").shift(-90).rolling_max(window_size=90, min_samples=30).over("subject").alias("GL_post_peak_90m"),
    pl.col("GL_interp").shift(-90).rolling_min(window_size=90, min_samples=30).over("subject").alias("GL_post_min_90m"),
]).with_columns([
    (pl.col("GL_post_peak_90m") - pl.col("GL_t0")).alias("delta_GL_peak_90m"),
    (pl.col("GL_post_mean_90m") - pl.col("GL_t0")).alias("delta_GL_mean_90m"),
    pl.col("Timestamp").dt.hour().alias("hour_of_day"),
])

# Filtra apenas os momentos de refeição para consolidar o novo dataset tabular
colunas_dataset = [
    "subject", "study_day", "Timestamp", "hour_of_day", "Meal Type",
    "Calories", "Carbs", "Protein", "Fat", "Fiber", "Amount Consumed",
    "GL_t0", "GL_pre_mean_90m", "GL_pre_std_90m", "GL_delta_pre_90m",
    "HR_pre_mean_90m", "METs_pre_sum_90m",
    "GL_post_mean_90m", "GL_post_peak_90m", "GL_post_min_90m",
    "delta_GL_peak_90m", "delta_GL_mean_90m"
]

df_novo = (
    df_janelas
    .filter(pl.col("Meal Type").is_not_null())
    .select(colunas_dataset)
    # padroniza rótulos: Breakfast/breakfast -> breakfast; Snack/Snacks/snack 1 -> snack
    .with_columns(pl.col("Meal Type").str.to_lowercase().str.replace(r"^snack.*$", "snack"))
)

# Cruza com o bio.csv (caso exista no diretório)
if BIO_PATH.exists():
    df_bio = pl.read_csv(BIO_PATH)
    df_bio = df_bio.rename({c: c.strip() for c in df_bio.columns})
    if "subject" not in df_bio.columns:
        df_bio = df_bio.rename({df_bio.columns[0]: "subject"})
    # só o perfil do participante (evita colunas de glicemia de dedo e "Time (t)_duplicated")
    df_bio = df_bio.select(["subject", "Age", "Gender", "BMI", "A1c PDL (Lab)", "Fasting GLU - PDL (Lab)"])
    df_novo = df_novo.join(df_bio, on="subject", how="left")

print(f"Novo dataset criado com shape: {df_novo.shape}")
display(df_novo.head(5))
```

```python
# Classe de diabetes pelo A1c (<5.7 saudável | 5.7-6.4 pré-diabético | >6.4 diabético)
classe = (
    df_bio.select("subject", pl.col("A1c PDL (Lab)").alias("a1c"))
    .with_columns(
        pl.when(pl.col("a1c") < 5.7).then(pl.lit("saudavel"))
        .when(pl.col("a1c") <= 6.4).then(pl.lit("pre_diabetico"))
        .otherwise(pl.lit("diabetico"))
        .alias("classe")
    )
)
df_novo = df_novo.join(classe.select("subject", "classe"), on="subject", how="left")

# Split por PARTICIPANTE, estratificado por classe (mesma pessoa nunca fica em treino e teste)
SEED, TEST_FRAC = 42, 0.2
part = classe.with_columns(
    pl.int_range(pl.len()).shuffle(seed=SEED).over("classe").alias("_rank"),
    (pl.len().over("classe") * TEST_FRAC).round().cast(pl.Int64).alias("_n_test"),
)
test_ids = part.filter(pl.col("_rank") < pl.col("_n_test"))["subject"]
is_test = pl.col("subject").is_in(test_ids.implode())

print("Participantes por classe e conjunto:")
display(
    part.with_columns(is_test.alias("teste"))
    .group_by("classe", "teste").len().sort("classe", "teste")
)
```

```python
import numpy as np

# Alvo: pico de glicose (mg/dL) nos 90 min seguintes à refeição (valor absoluto, sem depender de GL_t0)
TARGET = "GL_post_peak_90m"

# Features finais, separadas por papel no sistema de recomendação
FEATURES_DECISAO = {                      # o algoritmo genético escolhe
    "num": ["hour_of_day", "Carbs", "Protein", "Fat", "Fiber"],
    "cat": ["Meal Type"],
}
FEATURES_PACIENTE = {                     # dado fixo do paciente (disponível para um paciente novo)
    "num": ["Age", "BMI", "A1c PDL (Lab)", "Fasting GLU - PDL (Lab)"],
    "cat": ["Gender"],
}
# Fora de propósito: Calories (redundante com os macros), Amount Consumed (F~0),
# GL_t0 / GL_pre_* / HR_pre_* / METs_pre_* (estado antes da refeição, desconhecido num plano futuro)
# e GL_post_* / delta_* (alvo -> vazamento).

num  = FEATURES_DECISAO["num"] + FEATURES_PACIENTE["num"]
cats = FEATURES_DECISAO["cat"] + FEATURES_PACIENTE["cat"]

df_m = (
    df_novo.select(["subject", "classe"] + cats + num + [TARGET])
    .filter(pl.col(TARGET).is_not_null())
)
train, test = df_m.filter(~is_test), df_m.filter(is_test)
print(f"treino: {train.height} linhas / {train['subject'].n_unique()} participantes | "
      f"teste: {test.height} linhas / {test['subject'].n_unique()} participantes")

# imputação (mediana), padronização e one-hot: estatísticas calculadas SÓ no treino
med = train.select([pl.col(c).median() for c in num]).row(0, named=True)
mu  = train.select([pl.col(c).fill_null(med[c]).mean() for c in num]).row(0, named=True)
sd  = train.select([pl.col(c).fill_null(med[c]).std()  for c in num]).row(0, named=True)
levels = {c: sorted(train[c].drop_nulls().unique().to_list()) for c in cats}
levels["Gender"] = levels["Gender"][1:]   # drop_first: F/M seriam colunas redundantes

def prep(d):
    d = d.with_columns([((pl.col(c).fill_null(med[c]) - mu[c]) / sd[c]).alias(c) for c in num])
    d = d.with_columns([
        (pl.col(c) == v).cast(pl.Float32).alias(f"{c}={v}") for c in cats for v in levels[c]
    ])
    cols = num + [f"{c}={v}" for c in cats for v in levels[c]]
    return d.select(cols).to_numpy(), d[TARGET].to_numpy(), cols

X_train, y_train, cols = prep(train)
X_test,  y_test,  _    = prep(test)
print("X_train", X_train.shape, "| X_test", X_test.shape, "| NaN:", bool(np.isnan(X_train).any()))
print("\nFeatures finais (", len(cols), "):")
for papel, f in [("decisão (GA)", FEATURES_DECISAO), ("paciente", FEATURES_PACIENTE)]:
    print(f"  {papel:13s}:", f["num"] + [f"{c}=*" for c in f["cat"]])
```

```python
from sklearn.feature_selection import SelectKBest, f_regression

# Diagnóstico: ranking de f_regression ajustado SÓ no treino. Não corta nenhuma feature:
# as variáveis de decisão (macros, tipo e horário) precisam ficar no modelo para guiar o GA.
selector = SelectKBest(score_func=f_regression, k="all").fit(X_train, y_train)
ranking = pl.DataFrame({
    "feature": cols, "F": selector.scores_, "p": selector.pvalues_,
}).sort("F", descending=True)
with pl.Config(tbl_rows=30):
    display(ranking)

# Lista final de features usada pelo modelo
features_selecionadas = cols
print("Features selecionadas:", features_selecionadas)
print("X_train", X_train.shape, "| X_test", X_test.shape)
```
